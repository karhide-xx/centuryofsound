# Create Century_select.json from century.xlsx

This notebook creates the Century_select.json file from the main century.xlsx spreadsheet with enhanced metadata for the Century of Sound web player.

## What This Does

Transforms the century.xlsx spreadsheet into a JSON file with:
- **city_mp3**: MP3 filename based on city_id (e.g., "8837004.mp3")
- **memory_mp3**: MP3 filename based on memory_id (e.g., "8837005.mp3")
- **City artist**: Sound recordist name
- **Category**: Cultural/historical category (e.g., "Cultural practices")
- **Format**: Original recording format (e.g., "5-inch reel tape")
- All original metadata fields (titles, descriptions, country, year, etc.)

## Prerequisites
- Python 3.x
- century.xlsx file in "Supporting files" folder

## Output
- Century_select.json in the root directory

## Step 1: Install Dependencies

Install required packages:
- `pandas`: Data manipulation and analysis library
- `openpyxl`: Excel file reader/writer for .xlsx files

```python
import pandas as pd
import json
```

## Step 2: Import Libraries

Import necessary modules for Excel processing and JSON output.

```python
# Read the main century.xlsx file
century_excel_path = "Supporting files/century.xlsx"
output_json_path = "Century_select.json"

# Read Excel file into DataFrame
df_century = pd.read_excel(century_excel_path)

print(f"Read {len(df_century)} rows from century.xlsx")
print(f"\nColumns: {list(df_century.columns)}")
print(f"\nFirst row preview:")
print(df_century.head(1))
```

## Step 3: Read century.xlsx

Load the Excel file and display its structure to verify the data.

```python
# Transform the data to create the Century_select.json structure
records = []

for _, row in df_century.iterrows():
    # Helper function to convert NaN to None
    def get_value(key):
        val = row.get(key)
        return None if pd.isna(val) else val
    
    # Create the record with all required fields
    record = {
        "Number": get_value('Number'),
        "Order for player": get_value('Order for player'),  # Add Order for player field
        "City title": get_value('City title'),
        "City artist": get_value('Sound recordist'),  # Use Sound recordist as City artist
        "City description": get_value('City description'),
        "Memory title": get_value('Memory title'),
        "Memory artist": get_value('Memory artist'),
        "Memory description": get_value('Memory description'),
        "Country": get_value('Country'),
        "Cultural group": get_value('Cultural group'),  # Now properly converts NaN to None
        "Year": get_value('Year'),
        "Type ": get_value('Type ') if 'Type ' in df_century.columns else None,
        "Category": get_value('Category'),  # Add Category field
        "Format": get_value('Format'),  # Add Format field
        "city_mp3": f"{row.get('city_id')}.mp3" if pd.notna(row.get('city_id')) else None,  # MP3 filename from city_id
        "memory_mp3": f"{row.get('memory_id')}.mp3" if pd.notna(row.get('memory_id')) else None  # MP3 filename from memory_id
    }
    
    records.append(record)

# Sort records by Order for player field in ascending order (1 first)
records = sorted(records, key=lambda x: x.get('Order for player') if x.get('Order for player') is not None else float('inf'))

print(f"\n✓ Transformed {len(records)} records")
print(f"\nSample record:")
print(json.dumps(records[0], indent=2, ensure_ascii=False))
```

## Step 4: Transform Data

Transform the data to include:
- **city_mp3** and **memory_mp3**: MP3 filenames generated from city_id and memory_id
- **City artist**: Mapped from Sound recordist field
- **Category**: Cultural/historical category from the spreadsheet
- **Format**: Original recording format
- All original metadata fields preserved

```python
# Save to JSON file with pretty formatting
with open(output_json_path, 'w', encoding='utf-8') as f:
    json.dump(records, f, indent=2, ensure_ascii=False)

print(f"\n✓ Successfully created Century_select.json")
print(f"✓ Saved to: {output_json_path}")
print(f"✓ Total records: {len(records)}")
print(f"\nNew fields added:")
print("  - Category: Cultural/historical category")
print("  - Format: Original recording format")
print("  - city_mp3: MP3 filename based on city_id")
print("  - memory_mp3: MP3 filename based on memory_id")
print("  - City artist: Sound recordist name")
```

## Step 5: Save to Century_select.json

Save the transformed data to Century_select.json in the root directory with:
- Pretty-print formatting (2-space indentation)
- UTF-8 encoding with Unicode characters preserved
- Ready for use by the web player application
